# NeuroGolf 6645.39 Public Score Open Solution

Public score: **6645.39**.

This solution uses compact per-task ONNX graphs with static shapes and positive-cost wrappers where they reduce the metric while preserving public-example behavior. The artifact is bundled as a reproducible submission zip.

If this notebook saves you time or provides a useful benchmark, please upvote to support it.

```python
from pathlib import Path
import hashlib
import shutil
import zipfile

root = Path('/kaggle/input/neurogolf-6645-39-open-submission-artifact')
artifact = root / 'submission.zip'
out = Path('/kaggle/working/submission.zip')
if artifact.exists():
    shutil.copyfile(artifact, out)
else:
    onnx_dir = root / 'submission'
    with zipfile.ZipFile(out, 'w', zipfile.ZIP_DEFLATED) as zf:
        for path in sorted(onnx_dir.glob('task*.onnx')):
            zf.write(path, arcname=path.name)

with zipfile.ZipFile(out) as zf:
    onnx_files = sorted(name for name in zf.namelist() if name.startswith('task') and name.endswith('.onnx'))

print(f'Wrote {out}')
print(f'ONNX files: {len(onnx_files)}')
print(f'SHA256: {hashlib.sha256(out.read_bytes()).hexdigest()}')
```
