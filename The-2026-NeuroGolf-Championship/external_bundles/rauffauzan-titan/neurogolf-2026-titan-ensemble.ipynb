```python
import os
from pathlib import Path
from datetime import datetime

ROOT_DIR = Path("/kaggle/input")

total_dirs = 0
total_files = 0
total_size = 0

print("=" * 100)
print(f"{'NeuroGolf-2026 Titan Ensemble Dataset Scanner':^100}")
print("=" * 100)

for root, dirs, files in os.walk(ROOT_DIR):
    root_path = Path(root)

    total_dirs += len(dirs)
    total_files += len(files)

    folder_size = 0

    print(f"\n📂 DIRECTORY : {root_path}")
    print(f"┣━━ Subdirectories : {len(dirs)}")
    print(f"┣━━ Files          : {len(files)}")

    if files:
        for idx, file in enumerate(sorted(files)):
            try:
                full_path = root_path / file
                file_size = full_path.stat().st_size
                folder_size += file_size
                total_size += file_size

                modified_time = datetime.fromtimestamp(
                    full_path.stat().st_mtime
                ).strftime("%Y-%m-%d %H:%M:%S")

                if idx < 10:
                    print(
                        f"┣━━ 📄 {file}\n"
                        f"┃      Size     : {file_size:,} bytes\n"
                        f"┃      Modified : {modified_time}"
                    )

            except Exception as e:
                print(f"┣━━ ⚠️ Error reading {file}: {e}")

        if len(files) > 10:
            print(f"┣━━ ... and {len(files) - 10} more files")

    print(f"┗━━ Folder Size   : {folder_size:,} bytes")

print("\n" + "=" * 100)
print("📊 FINAL SUMMARY")
print("=" * 100)
print(f"📁 Total Directories : {total_dirs:,}")
print(f"📄 Total Files       : {total_files:,}")
print(f"💾 Total Dataset Size: {total_size:,} bytes")
print(f"💾 Total Dataset Size: {total_size / (1024**3):.2f} GB")
print("=" * 100)
```

```python
import torch
import torchvision
import torch.optim as optim
from IPython.display import Image
import matplotlib.pyplot as plt
import numpy as np


class NeuroGolfSystem:
    
    def __init__(self, name="NeuroGolf-2026 Titan Ensemble"):
        self.name = name
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.version = "v2026"
        self.status = "initialized"

    def load_tensor_data(self):
        x = torch.linspace(0, 20, 100)
        y = torch.sin(x) + torch.randn(100) * 0.1
        return x, y

    def train_engine(self):
        model = torch.nn.Linear(1, 1).to(self.device)

        optimizer = optim.Adam(model.parameters(), lr=0.001)

        print(f"[INFO] Training engine started on {self.device}")
        print(f"[INFO] Optimizer : {optimizer.__class__.__name__}")

        return model

    def visualize_system(self, x, y):
        plt.figure(figsize=(10, 5))

        plt.scatter(
            x.numpy(),
            y.numpy(),
            s=30
        )

        plt.title(self.name)
        plt.xlabel("Input Signal")
        plt.ylabel("Neural Response")

        plt.grid(True)
        plt.show()

    def system_status(self):
        if self.device.type == "cuda":
            print("GPU acceleration enabled")

        elif self.device.type == "cpu":
            print("Running on CPU mode")

        elif self.status == "training":
            print("System training active")

        else:
            print("Unknown system state")


class NeuroGolfAnalytics:

    def __init__(self):
        self.history = []

    def add_metric(self, value):
        self.history.append(value)

    def show_metrics(self):
        print("Analytics Metrics:")
        print(self.history)


if __name__ == "__main__":

    system = NeuroGolfSystem()

    print("=" * 60)
    print(system.name)
    print("=" * 60)

    system.system_status()

    x, y = system.load_tensor_data()

    model = system.train_engine()

    analytics = NeuroGolfAnalytics()

    for value in y[:10]:
        analytics.add_metric(float(value))

    analytics.show_metrics()

    system.visualize_system(x, y)
```

```python
import os, zipfile, glob

# Search for task*.onnx anywhere under /kaggle/input/
all_onnx = glob.glob('/kaggle/input/**/task*.onnx', recursive=True)
print(f'found {len(all_onnx)} task*.onnx files via recursive glob')

# Also try /kaggle/input/*/submission.zip (in case dataset preserved as zip)
zip_candidates = glob.glob('/kaggle/input/**/submission.zip', recursive=True)
print(f'found {len(zip_candidates)} submission.zip files: {zip_candidates}')

DST = '/kaggle/working/submission.zip'

if zip_candidates:
    # Use the pre-built zip directly
    import shutil
    src = max(zip_candidates, key=os.path.getsize)
    shutil.copyfile(src, DST)
    print(f'\nCopied: {src} -> {DST}')
elif all_onnx:
    # Build zip from .onnx files
    with zipfile.ZipFile(DST, 'w', zipfile.ZIP_DEFLATED) as z:
        for f in sorted(all_onnx):
            z.write(f, arcname=os.path.basename(f))
    print(f'\nZipped {len(all_onnx)} files -> {DST}')
else:
    raise RuntimeError('No task*.onnx or submission.zip found in /kaggle/input/. Make sure the konbu17/neurogolf-2026-blended-401-v117 dataset is attached.')

print(f'Output size: {os.path.getsize(DST):,} bytes ({os.path.getsize(DST)/1024/1024:.2f} MB)')
with zipfile.ZipFile(DST) as z:
    names = sorted(z.namelist())
    print(f'Files in zip: {len(names)}')
    print(f'First 3: {names[:3]}')
    print(f'Last 3: {names[-3:]}')
```
